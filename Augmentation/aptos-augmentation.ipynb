```python
import os
import shutil
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Parameters
image_size = 299
batch_size = 16
target_num_images = 1200

def augment_images(input_folder, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Copy original images first
    for file_name in os.listdir(input_folder):
        source = os.path.join(input_folder, file_name)
        target = os.path.join(output_folder, file_name)
        if not os.path.exists(target):
            shutil.copyfile(source, target)
    
    current_num_images = len(os.listdir(output_folder))
    remaining_images = target_num_images - current_num_images
    print(f"Original: {current_num_images} | Need {remaining_images} more")
    
    if remaining_images <= 0:
        print("Already enough images, skipping.")
        return

    # ✅ NO rescale
    datagen = ImageDataGenerator(
        rotation_range=180,
        width_shift_range=0.1,
        height_shift_range=0.1,
        zoom_range=0.1,
        horizontal_flip=True,
        vertical_flip=True,
        fill_mode='reflect'
    )

    aug_datagen = datagen.flow_from_directory(
        os.path.dirname(output_folder),
        save_to_dir=output_folder,
        save_format='png',
        target_size=(image_size, image_size),
        batch_size=batch_size,
        classes=[os.path.basename(output_folder)],
    )

    num_batches = int(np.ceil(remaining_images / batch_size))
    for i in range(num_batches):
        next(aug_datagen)
        if i % 100 == 0:
            print(f"Progress: {i}/{num_batches} batches done...")

    print(f"✅ Done! Total images: {len(os.listdir(output_folder))}")


# ===== Mild DR only =====
input_path  = '/kaggle/input/datasets/humayratasnim/before-augmented-messidor-2/Messidor-2/4_Proliferative_DR'
output_path = '/kaggle/working/4_Proliferative_DR'

augment_images(input_path, output_path)

# Zip it
shutil.make_archive('/kaggle/working/4_Proliferative_DR', 'zip', '/kaggle/working/4_Proliferative_DR')
print("✅ Zipped: 4_Proliferative_DR.zip — ready to download!")
```
